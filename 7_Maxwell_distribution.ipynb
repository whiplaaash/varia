```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
Nx = 100          # spatial grid points
Nv = 100          # velocity grid points
Lx = 1.0          # domain length
Vmax = 5.0
tau = 0.1         # relaxation time
dt = 0.001
t_final = 0.2

# Grids
x = np.linspace(0, Lx, Nx)
v = np.linspace(-Vmax, Vmax, Nv)
dx = x[1] - x[0]
dv = v[1] - v[0]

# Initial distribution f(x,v)
X, V = np.meshgrid(x, v, indexing='ij')
f = np.exp(-V**2) * (1 + 0.1 * np.sin(2*np.pi*X/Lx))

def maxwellian(n, u, T):
    """Return Maxwellian distribution."""
    return n / np.sqrt(2*np.pi*T) * np.exp(-(V - u)**2 / (2*T))
```
